```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

name = []
cp = []
pp = []
hp = []

for page in range(1,50):  

    if page == 1:
        url = "https://steamcharts.com/top"
    else:
        url = f"https://steamcharts.com/top/p.{page}"

    response = requests.get(
        url,
        headers={
            "User-Agent": "Mozilla/5.0"
        }
    )

    soup = BeautifulSoup(response.text, "lxml")

    rows = soup.select("table tbody tr")

    for row in rows:

        game = row.select_one("td.game-name")
        nums = row.select("td.num")

        if game and len(nums) >= 3:

            name.append(game.text.strip())

            cp.append(
                nums[0].text.strip().replace(",", "")
            )

            pp.append(
                nums[1].text.strip().replace(",", "")
            )

            hp.append(
                nums[2].text.strip().replace(",", "")
            )

df = pd.DataFrame({
    "Name": name,
    "Current_Players": cp,
    "Peak_Players": pp,
    "Hours_Played": hp
})


for col in ["Current_Players", "Peak_Players", "Hours_Played"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.head())
print(df.shape)


df.to_csv("steamcharts_top_games.csv", index=False)
```
