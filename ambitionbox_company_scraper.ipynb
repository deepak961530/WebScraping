```python
import pandas as pd
import requests
from bs4 import BeautifulSoup 
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'}

webpage=requests.get(
    'https://www.ambitionbox.com/list-of-companies?ratings=4.5&sortBy=popular&locations=new-delhi,mumbai,bengaluru,hyderabad,pune,chennai,kolkata,gurgaon,noida,ahmedabad',
    headers=headers
).text
```

```python
soup=BeautifulSoup(webpage,'lxml')
```

```python
company=soup.find_all('div',class_='companyCardWrapper')
```

```python
name=[]
rating=[]
review=[]
ct=[]
location=[]
salaries=[]
interviews=[]
jobs=[]
benifits=[]
photos=[]
hrt=[]
crt=[]
for i in company:
    name.append(i.find('h2','companyCardWrapper__companyName').text.strip())
    rating.append(i.find('div','rating_star_container').text.strip())
    review.append(i.find('span','companyCardWrapper__companyRatingCount').text.strip())
    temp=(i.find('span','companyCardWrapper__interLinking').text.strip())
    temp = i.find('span','companyCardWrapper__interLinking').text.strip()

    parts = temp.split('|')

    ct.append(parts[0].strip())

    if len(parts) > 1:
        location.append(parts[1].strip())
    else:
        location.append(None)
    temp3=i.find_all('span','companyCardWrapper__ActionCount')
    salaries.append(temp3[1].text.strip())
    interviews.append(temp3[2].text.strip())
    jobs.append(temp3[3].text.strip())
    benifits.append(temp3[4].text.strip())
    photos.append(temp3[5].text.strip())
    temp5=i.find('span','companyCardWrapper__ratingHeader--high')
    if temp5:
        temp4=i.find('span','companyCardWrapper__ratingValues')
        hrt.append(temp4.text.strip())
    else:
        hrt.append(None)
    temp6=i.find('span','companyCardWrapper__ratingHeader--critical')
    if temp6:
        temp7=i.find('span','companyCardWrapper__ratingValues')
        crt.append(temp7.text.strip())
    else:
        crt.append(None)
    
```

```python
df = pd.DataFrame({

    'Company_Name': name,
    'Rating': rating,
    'Reviews': review,
    'Company_Type': ct,
    'Location': location,
    'Salaries': salaries,
    'Interviews': interviews,
    'Jobs': jobs,
    'Benefits': benifits,
    'Photos': photos,
    'Highly_Rated_For': hrt,
    'Critically_Rated_For': crt

})

df
```
